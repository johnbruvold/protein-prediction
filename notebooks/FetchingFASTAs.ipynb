Now that I have the proteins from Uniprot, I need to clean the proteins before creating the model. Specifically, I want to avoid training on small proteins(<50 amino acids) as well as ones with non-canonical amino acids. 

```python
from pathlib import Path
import pandas as pd
proj_path = Path.cwd().parent

hydrolases_raw = pd.read_csv(proj_path / "data" / "RawSequences" / "Hydrolases.csv")
transferases_raw = pd.read_csv(proj_path / "data" / "RawSequences" / "Transferases.csv")

#Remove small protein sequences and those with non-canonical AAs
hydrolases_raw = hydrolases_raw[(hydrolases_raw["Length"] >= 50) &  ~(hydrolases_raw["Sequence"].str.contains("[OUXB]",regex=True))]
transferases_raw = transferases_raw[(transferases_raw["Length"] >= 50) &  ~(transferases_raw["Sequence"].str.contains("[OUXB]",regex=True))]

#Remove any duplicate sequences
hydrolases_raw = hydrolases_raw.drop_duplicates("Sequence")
transferases_raw = transferases_raw.drop_duplicates("Sequence")

hydrolases_raw.to_csv(proj_path / "data" / "CleanedSequences" / "HydrolasesCleaned.csv")
transferases_raw.to_csv(proj_path / "data" / "CleanedSequences" / "TransferasesCleaned.csv")
```
