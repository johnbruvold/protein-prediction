Now that I have the proteins from Uniprot, I need to clean the proteins before creating the model. Specifically, I want to avoid training on small proteins(<50 amino acids) as well as ones with non-canonical amino acids. 

```python
from pathlib import Path
import pandas as pd
import os

proj_path = Path.cwd().parent
csv_files = os.listdir(proj_path / "data" / "RawSequences" )
csv_files = [file for file in csv_files if ".csv" in file] 
for csv in csv_files:
    raw_file = pd.read_csv(proj_path / "data" / "RawSequences" / f"{csv[:-4]}.csv")


    #Remove small protein sequences and those with non-canonical AAs
    raw_file = raw_file[(raw_file["Length"] >= 50) &  ~(raw_file["Sequence"].str.contains("[OUXB]",regex=True))]

    #Remove any duplicate sequences
    raw_file = raw_file.drop_duplicates("Sequence")


    raw_file.to_csv(proj_path / "data" / "CleanedSequences" / f"{csv[:-4]}Cleaned.csv")
```
